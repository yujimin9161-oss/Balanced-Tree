##09

삽입된 새 노드 $p$(Red)의 부모 $p^1$도 Red인 상황에서, $p^1$의 형제(즉 삼촌 노드)도 Red일 때 색상 반전(Recoloring)이 일어난다.

1. 부모 $p^1$과 삼촌 노드를 Black으로 칠하고, 조부모 $p^2$를 Red로 변경한다.
2. 이제 $p^2$가 Red가 되었는데 증조부모 $p^3$도 Red라면 다시 성질 3 위반(Red-Red)이 발생한다.
3. 이때 $p^3$의 형제 노드도 Red라면 또 다시 Recoloring이 일어나 $p^3$와 그 형제를 Black으로, 고조부모 $p^4$를 Red로 칠한다.
4. 다시 $p^4$가 Red가 되어 $p^5$(Red)와 충돌하고, $p^5$의 형제가 Red라면 $p^5$와 형제를 Black으로, $p^6$를 Red로 반전시킨다.이처럼 삼촌 노드가 계속해서 Red인 조건이 맞아떨어지면, 조상 노드들로 올라가며 연속적으로 Recoloring 연산이 루트 노드($p^n$)까지 연쇄적으로 파급되어 영향을 미칠 수 있다.

노드가 가질 수 있는 최대 키가 1,000개이면 트리의 차수(M)는 1,001차이다.최소 높이: 모든 노드가 키를 최대(1000개)로 꽉 채웠을 때. $1001^h - 1 \ge 10^{12}$ 이므로, 대략 $h = 4$.최대 높이: 루트를 제외한 모든 노드가 키를 최소치(500개)만 가질 때. 빈약한 트리일수록 높이가 높아진다. 이 조건으로 계산하면 $h = 6$.따라서 B-트리의 높이 범위는 4 ~ 6 이다.
